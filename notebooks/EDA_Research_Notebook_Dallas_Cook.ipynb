## Univariate Analysis

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Identify numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns

for col in numerical_cols:
    plt.figure(figsize=(14, 5))

    # Histogram
    plt.subplot(1, 2, 1)
    sns.histplot(df[col], kde=True, bins=15)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel("Frequency")

    # Boxplot
    plt.subplot(1, 2, 2)
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.xlabel(col)

    plt.tight_layout()
    plt.show()

    # Outlier detection (IQR method)
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"\nOutlier Detection for {col}:")
    print(f"Lower Bound = {lower_bound:.2f}, Upper Bound = {upper_bound:.2f}")
    print(f"Number of Outliers: {outliers.shape[0]}")
    print("-" * 50)
```

```python
import scipy.stats as stats

# Filter data for writing tasks only
writing_df = df[df['TaskType'] == 'Writing']

# Group satisfaction ratings by AI level
group1 = writing_df[writing_df['AI_AssistanceLevel'] == 1]['SatisfactionRating']
group2 = writing_df[writing_df['AI_AssistanceLevel'] == 2]['SatisfactionRating']
group3 = writing_df[writing_df['AI_AssistanceLevel'] == 3]['SatisfactionRating']

# Perform one-way ANOVA
f_stat, p_value = stats.f_oneway(group1, group2, group3)

print("ANOVA Results:")
print("F-Statistic:", round(f_stat, 3))
print("p-Value:", round(p_value, 5))

# Support with boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(data=writing_df, x='AI_AssistanceLevel', y='SatisfactionRating')
plt.title('Satisfaction Rating by AI Assistance Level (Writing Tasks)')
plt.xlabel('AI Assistance Level')
plt.ylabel('Satisfaction Rating')
plt.tight_layout()
plt.show()
```

```python
# Create contingency table
contingency_table = pd.crosstab(writing_df['AI_AssistanceLevel'], writing_df['FinalOutcome'])

# Chi-Square Test
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

print("Chi-Square Test Results:")
print("Chi-Square Statistic:", round(chi2, 3))
print("Degrees of Freedom:", dof)
print("p-Value:", round(p, 5))

# Support with stacked bar plot
normalized_table = contingency_table.div(contingency_table.sum(axis=1), axis=0)
normalized_table.plot(kind='bar', stacked=True, colormap='viridis', figsize=(8, 5))
plt.title('Assignment Outcome by AI Assistance Level (Writing Tasks)')
plt.xlabel('AI Assistance Level')
plt.ylabel('Proportion of Outcomes')
plt.legend(title='Final Outcome')
plt.tight_layout()
plt.show()
```

# Discussion Notes

1. Interpretation
The results of this study demonstrate that higher levels of AI assistance during writing tasks are significantly associated with increased student satisfaction and a higher likelihood of assignment completion. In practical terms, this means that when students receive more substantial support from AI tools — such as advanced writing suggestions or guided task feedback — they are more likely to feel confident, engaged, and capable of completing their academic work.

These findings directly answer the research question: “Does a higher level of AI assistance lead to greater satisfaction and assignment completion in writing tasks?” The statistical significance of both the ANOVA (p = 0.0009) and Chi-square test (p = 0.0009) confirms that AI support plays a meaningful role in shaping academic outcomes.

An alternative explanation could be that students who are already motivated or academically strong are more likely to engage deeply with AI tools, leading to better outcomes. Without real user traits or demographic controls, we cannot fully rule out this possibility.

2. Comparison with Literature
These findings align closely with prior research. Kizilcec et al. (2024) found that students using AI assistants demonstrated improved academic performance and motivation. Similarly, Zheldibayeva et al. (2025) showed that tailored AI feedback enhanced writing quality in high school students. Our results support these conclusions and extend them by showing that increased AI assistance not only improves satisfaction but also correlates with actual task completion.

Additionally, Liu et al. (2025) emphasized the role of AI literacy and self-regulation in maximizing AI effectiveness. Our findings indirectly support this by suggesting that students benefit most when they engage meaningfully with higher levels of AI support. The new insight our study offers is a quantitative, usage-based view — drawn from session-level interaction data — that connects intensity of AI use to measurable outcomes.

3. Limitations
One of the key limitations of this study is the use of a synthetic dataset. While it was well-structured and designed to mimic real-world behavior, it does not reflect the full complexity of human learning, emotional factors, or motivation. Additionally, the dataset lacks demographic variables, making it impossible to explore how factors like age, gender, or academic background influence AI engagement.

Methodologically, the analysis was limited to observational comparisons and basic statistical tests (ANOVA and Chi-square). Without experimental design or real behavioral tracking, causal conclusions cannot be drawn.

4. Implications
These findings suggest that educational institutions and developers of AI tools should consider offering higher levels of interactive, guided AI support, especially in writing-intensive courses. Students appear to gain more satisfaction and achieve better outcomes when AI tools are integrated as substantial collaborators rather than light-touch aids.

Stakeholders who should pay attention to these results include:

Educators, who can use AI to scaffold learning

EdTech companies, to design adaptive and supportive AI tools

Curriculum designers, to integrate AI use into writing tasks in meaningful ways

### Abstract

This study explores whether higher levels of AI assistance influence student satisfaction and assignment completion during writing tasks. The central research question asks: Does a higher level of AI support lead to improved academic outcomes in student writing sessions?

Using a synthetic dataset simulating AI assistant usage among students, we conducted univariate and bivariate analyses, followed by an ANOVA and Chi-square test to assess relationships between assistance level, satisfaction rating, and task completion.

Results showed that students receiving higher levels of AI assistance reported significantly greater satisfaction (p = 0.0009) and were more likely to complete their assignments (p = 0.0009). Visualizations, including boxplots and stacked bar charts, further supported these outcomes.

These findings highlight the potential of robust AI support systems to enhance learning experiences and academic performance. Future research should expand to real-world data, explore demographic factors, and examine long-term impacts on student achievement.
